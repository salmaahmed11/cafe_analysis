```python
data['hour'] = data['datetime'].dt.hour
data['day'] = data['datetime'].dt.day
data['month'] = data['datetime'].dt.month
data['year'] = data['datetime'].dt.year
def get_time_of_day(hour):
    if 5 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 21:
        return 'Evening'
    else:
        return 'Night'
data['time_of_day'] = data['hour'].apply(get_time_of_day)
data
```

```python
daily_sales = data.groupby("datetime")["money"].sum()
daily_sales.tail(100)
```

```python
daily_sales.tail(100).plot(kind='line', figsize=(10,5),color="brown")
plt.title("Daily Sales (Last 100 Days)")
plt.xlabel("Date")
plt.ylabel("Total Money")
plt.grid(True)
plt.show()
```

```python
hourly_sales = data.groupby("hour")["money"].sum().sort_values(ascending=False)
hourly_sales
```

```python
daily_sales = data.groupby("time_of_day")["money"].sum().sort_values(ascending=False)
daily_sales.head()
```

```python
daily_sales.plot(kind="bar", color=["orange", "gold", "purple", "blue"], figsize=(8,6))
plt.title("Total Sales by Time of Day")
plt.xlabel("Time of Day")
plt.ylabel("Total Money")
plt.xticks(rotation=0)
plt.show()
```

```python
top_coffees = data.groupby("coffee_name")["money"].sum().sort_values(ascending=False)
top_coffees
```

```python
top_coffees.plot(kind="bar", figsize=(10,6), color="saddlebrown")
plt.title("Top 10 Coffees by Sales")
plt.xlabel("Coffee Name")
plt.ylabel("Total Money")
plt.xticks(rotation=45)
plt.show()
```

```python
data['weekday'] = data['datetime'].dt.day_name()
weekday_sales = data.groupby("weekday")["money"].sum().sort_values(ascending=False)
weekday_sales
```

```python
# weekday_sales = weekday_sales.reindex(["Saturday","Sunday","Monday","Tuesday","Wednesday","Thursday","Friday"]).sort_values(ascending=False)
weekday_sales = weekday_sales.reindex(["Saturday","Sunday","Monday","Tuesday","Wednesday","Thursday","Friday"])
weekday_sales.plot(kind="bar", figsize=(8,6), color="brown")
plt.title("Total Sales by Weekday")
plt.xlabel("Weekday")
plt.ylabel("Total Money")
plt.xticks(rotation=0)
plt.show()
```

```python
pivot = data.pivot_table(values="money", index="weekday", columns="hour", aggfunc="sum")
pivot_normalized = pivot.div(pivot.sum(axis=1), axis=0).fillna(0)
plt.figure(figsize=(12,6))
sns.heatmap(pivot_normalized, cmap="YlOrBr", annot=True, fmt=".0%")
plt.title("Normalized Sales Heatmap by Hour and Weekday (Percentage)")
plt.xlabel("Hour")
plt.ylabel("Weekday")
plt.show()
```
